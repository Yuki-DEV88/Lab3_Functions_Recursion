```python
LAST_NAME = "Cardenas"
STUDENT_ID = "TUPM-26-1258"
FAVORITE_ARTIST = "Maroon 5"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

OPERATING_RANGES = {
    "Temperature": (15, 90),   
    "Pressure": (30, 140),    
    "Voltage": (100, 115),     
    "Vibration": (0, 8),       
    "Runtime": (0, 24),        
}

EXECUTION_LOG = []

def log_execution(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        EXECUTION_LOG.append(f"{func.__name__} executed -> result: {result}")
        return result
    return wrapper

def generate_equipment_data():
    ascii_sum_name = sum(ord(c) for c in LAST_NAME.upper())
    ascii_sum_artist = sum(ord(c) for c in FAVORITE_ARTIST.upper() if c.isalpha())

    readings = {
        "Temperature": (ascii_sum_name + SEED_DIGIT) % 60 + 10,
        "Pressure": (ascii_sum_artist + SEED_DIGIT * 3) % 120 + 20,
        "Voltage": (ID_SUM * SEED_DIGIT) % 15 + 100,
        "Vibration": abs(ascii_sum_name - ascii_sum_artist) % 10,
        "Runtime": "N/A",  # simulated malformed sensor entry
    }
    return readings

@log_execution
def validate_reading(name, value):
    try:
        numeric_value = float(value)
    except (ValueError, TypeError):
        return False, f"Invalid non-numeric value ('{value}')"

    low, high = OPERATING_RANGES[name]
    if not (low <= numeric_value <= high):
        return False, f"Out of range ({numeric_value})"

    return True, numeric_value


def process_validation(readings):
    valid_values = {}
    invalid_entries = {}

    for name, raw_value in readings.items():
        is_valid, result = validate_reading(name, raw_value)
        if is_valid:
            valid_values[name] = result
        else:
            invalid_entries[name] = result

    return valid_values, invalid_entries

@log_execution
def calculate_stats(*valid_values):
    if not valid_values:
        return {"total": 0, "average": 0, "max": None, "min": None}

    total = sum(valid_values)
    average = round(total / len(valid_values), 2)
    return {"total": total, "average": average,
            "max": max(valid_values), "min": min(valid_values)}

@log_execution
def classify_condition(invalid_count):
    if invalid_count == 0:
        return "OPTIMAL"
    elif invalid_count <= 1:
        return "WARNING"
    else:
        return "CRITICAL"

def display_summary(title, **stats):
    print(f"\n{title}")
    print("-" * 60)
    for key, value in stats.items():
        print(f"  {key}: {value}")


def display_report(readings, valid_values, invalid_entries, stats, condition):
    print("=" * 60)
    print(f"EQUIPMENT DIAGNOSTIC REPORT  (Engineer: {LAST_NAME}, "
          f"STUDENT_ID: {STUDENT_ID}, Favorite Artist: {FAVORITE_ARTIST})")
    print("=" * 60)

    print("\n1. GENERATED EQUIPMENT DATA")
    print("-" * 60)
    for name, raw_value in readings.items():
        print(f"  {name}: {raw_value}")

    print("\n2. VALIDATION RESULTS")
    print("-" * 60)
    print("  Valid readings:")
    for name, value in valid_values.items():
        print(f"    {name}: {value}  (range {OPERATING_RANGES[name]})")
    print("  Invalid readings:")
    for name, reason in invalid_entries.items():
        print(f"    {name}: {reason}")

    display_summary("3. DIAGNOSTIC RESULTS",
                     Total=stats["total"], Average=stats["average"],
                     Max=stats["max"], Min=stats["min"],
                     Condition=condition)

    print("\n4. EXECUTION LOG")
    print("-" * 60)
    for entry in EXECUTION_LOG:
        print(f"  {entry}")

    print("\n5. FINAL OUTPUT")
    print("-" * 60)
    print(f"  Equipment condition: {condition} "
          f"({len(invalid_entries)} invalid reading(s) out of {len(readings)})")
    print("=" * 60)


def main():
    readings = generate_equipment_data()
    valid_values, invalid_entries = process_validation(readings)
    stats = calculate_stats(*valid_values.values())
    condition = classify_condition(len(invalid_entries))
    display_report(readings, valid_values, invalid_entries, stats, condition)


if __name__ == "__main__":
    main()
```

```python
LAST_NAME = "Cardenas"
STUDENT_ID = "TUPM-26-1258"
FAVORITE_ARTIST = "Maroon 5"

SEED_DIGIT = int(STUDENT_ID[-1])

EXECUTION_LOG = []

def generate_fault_code():
    """
    Build a unique, reproducible multi-digit fault code from the ASCII
    values of LAST_NAME and FAVORITE_ARTIST, combined with SEED_DIGIT.
    """
    ascii_sum_name = sum(ord(c) for c in LAST_NAME.upper())
    ascii_sum_artist = sum(ord(c) for c in FAVORITE_ARTIST.upper() if c.isalpha())

    fault_code = (ascii_sum_name * SEED_DIGIT) + ascii_sum_artist
    EXECUTION_LOG.append(
        f"Fault code generated: ({ascii_sum_name} * {SEED_DIGIT}) + "
        f"{ascii_sum_artist} = {fault_code}"
    )
    return fault_code

def fault_trace(code, level=1, trace_log=None, call_counter=None):
    if trace_log is None:
        trace_log = []
    if call_counter is None:
        call_counter = [0]

    call_counter[0] += 1
    digit_sum = sum(int(d) for d in str(code))
    trace_log.append((level, code, digit_sum))

    if digit_sum < 10:
        return digit_sum, trace_log, call_counter[0]

    return fault_trace(digit_sum, level + 1, trace_log, call_counter)

def display_report(fault_code, final_result, trace_log, call_count):
    print("=" * 60)
    print(f"RECURSIVE FAULT TRACE REPORT  (Engineer: {LAST_NAME}, "
          f"STUDENT_ID: {STUDENT_ID}, Favorite Artist: {FAVORITE_ARTIST})")
    print("=" * 60)

    print("\n1. GENERATED FAULT DATA")
    print("-" * 60)
    print(f"  Fault Code: {fault_code}")

    print("\n2. RECURSIVE TRACE")
    print("-" * 60)
    for level, code, digit_sum in trace_log:
        print(f"  Level {level}: code={code} -> digit_sum={digit_sum}")

    print("\n3. NUMBER OF RECURSIVE CALLS")
    print("-" * 60)
    print(f"  {call_count}")

    print("\n4. EXECUTION LOG")
    print("-" * 60)
    for entry in EXECUTION_LOG:
        print(f"  {entry}")
    print(f"  Recursion terminated at digit_sum={final_result} "
          f"(base condition: single digit, < 10)")

    print("\n5. FINAL OUTPUT")
    print("-" * 60)
    print(f"  Fault Code {fault_code} traced through {call_count} "
          f"recursive call(s) -> Final Result: {final_result}")
    print("=" * 60)


def main():
    fault_code = generate_fault_code()
    final_result, trace_log, call_count = fault_trace(fault_code)
    display_report(fault_code, final_result, trace_log, call_count)


if __name__ == "__main__":
    main()
```
